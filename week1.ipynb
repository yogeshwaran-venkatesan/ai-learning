```python
PURITY_FACTORS = {
'22K': 0.916,
'18K': 0.75,
'24K': 0.999
}
def fine_gold(weight,purity):
    return weight * PURITY_FACTORS[purity]

print(fine_gold(10, "22K"))   # should print 9.16
```
