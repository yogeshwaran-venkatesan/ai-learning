```python
items = [
    {"code": "RG-1001", "metal": "Gold",   "purity": "22K", "weight": 8.45},
    {"code": "CH-2001", "metal": "Gold",   "purity": "22K", "weight": 24.10},
    {"code": "BG-3001", "metal": "Gold",   "purity": "18K", "weight": 18.30},
    {"code": "RG-1002", "metal": "Silver", "purity": "925", "weight": 6.20},
]

gold_items=[item for item in items if item['metal']=='Gold']
code_items = [item['code'] for item in items]
sum_Value=round(sum(item['weight'] for item in items if item['purity']=='22K'), 2)
print(f"gold_items: {gold_items}")
print(code_items)
print(f"Sum of weights: {sum_Value}")
```

```python
items = [
    {"code": "RG-1001", "metal": "Gold",   "purity": "22K", "weight": 8.45},
    {"code": "CH-2001", "metal": "Gold",   "purity": "22K", "weight": 24.10},
    {"code": "BG-3001", "metal": "Gold",   "purity": "18K", "weight": 18.30},
    {"code": "RG-1002", "metal": "Silver", "purity": "925", "weight": 6.20},
]
totals = {}
for item in items:
    if item['metal']=='Gold':
        purity = item.get('purity')
        weight = item['weight']
        if purity not in totals:
            totals[purity] = 0
        totals[purity]+=weight
print(totals)
```
