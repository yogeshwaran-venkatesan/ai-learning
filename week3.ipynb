```python

class CategoryCreator:
        def __init__(self):
            self.metals = ["Gold","Silver","Platinum"]
            self.metal_purity= {        
            'Gold':['22K', "18K", "24K", "14K"],
            "Silver":['925'],
            "Platinum":['950']
            }
            self.weights=[80, 15, 5]
            self.metal_category=["Ring", "Chain", "Bangle", "Necklace", "Earring"]
            self.branches=["Chennai", "Madurai", "Coimbatore", "Trichy"]
        def make_item(self):
                metal=random.choices(self.metals, weights=self.weights)[0]
                purity=random.choice(self.metal_purity[metal])
                category=random.choice(self.metal_category)
                weight_gm= round(random.uniform(2,60),2)
                branch=random.choice(self.branches)
                item_code = f"{category[:2].upper()}-{random.randint(1000, 9999)}"
                return { 
                 "metal":metal,
                 "purity":purity,
                 "weight_gm":weight_gm,
                 "branch":branch,
                 "item_code":item_code,
                 "category":category
                }

creator = CategoryCreator()
random.seed(42)
creator = CategoryCreator()


items = [creator.make_item() for _ in range(5000)]
print(len(items))
print(items[4444])
         
```
