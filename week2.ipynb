```python
PURITY_FACTORS = {"22K": 0.916, "18K": 0.75, "24K": 0.999}

# your class here
class JewelleryItem:
    def __init__(self,code,weight,purity):
        self.code=code
        self.weight= weight
        self.purity=purity
    def describe(self):
     return f"{self.code} weighs {self.weight} g ({self.purity})"
    def fine_gold(self):
        return self.weight * PURITY_FACTORS[self.purity]
chain = JewelleryItem("CH-2001", 24.10, "22K")
print(chain.describe())
print(round(chain.fine_gold(),2))    # expected: about 22.08
```

```python
items = [
    JewelleryItem("RG-1001", 8.45, "22K"),
    JewelleryItem("CH-2001", 24.10, "22K"),
    JewelleryItem("BG-3001", 18.30, "18K"),
]

# 1. loop through items and print each one's describe()
finegold =0
for item in items:
    desc = item.describe()
    print(f'{desc}')
# 2. calculate the TOTAL fine gold of all items (use sum + fine_gold())
    finegold += item.fine_gold()
print(f'Total fine gold:{round(finegold,2)}')
```

```python
import json
def parse_ai_output(text):
    # try / except here
    try:
        valid_json= json.loads(text)
        return valid_json
    except json.JSONDecodeError:
        return

print(parse_ai_output('{"metal": "Gold", "purity": "22K"}'))   # → {'metal': 'Gold', 'purity': '22K'}
print(parse_ai_output('{"metal": "Gold", "purity": '))          # → Invalid JSON from AI, then None
print(parse_ai_output('hello'))       
```
